```python
from simple_foraging_env import (
    SimpleForagingEnv, SimpleAgent, RandomAgent, GreedySwitchAgent,
    StalkerAgent, ChameleonAgent,
)
from opponent_model import OpponentModel
from q_agent import QLearningAgent, ReplayBuffer
from q_agent_classic import QLearningAgentClassic
from omg_args import OMGArgs
from maps import *
import transformers as t
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import matplotlib.patches as mpatches
import os
import glob
import random
import time
from IPython.display import clear_output
from collections import deque
import torch
from torch import nn
import torch.nn.functional as F
import numpy as np
import pandas as pd
import plotly.express as px
from sklearn.manifold import TSNE
```

```python
# ============================================================
# CONFIGURATION
# These must match the run that produced your checkpoints.
# Checkpoints come from simple_foraging_singleagent.py and are saved as
#   ./models/{FOLDER_ID}/{prefix}_ep{N}.pth
#   prefix in {classic_qnet, om_qnet, om_inference}
# ============================================================
FOLDER_ID = 0            # ./models/{FOLDER_ID}/...
MAP_NAME  = "MAP_4"      # one of MAP_1 .. MAP_5
OPPONENT  = "greedy"     # simple | greedy | stalker | chameleon
EPOCH     = 60         # int for a specific epoch, or None -> auto-pick latest
DEVICE    = "cpu"        # "cuda" / "mps" if available

map_layout = globals()[MAP_NAME]
device = DEVICE

# Opponent agent (must match the one trained against)
_OPPONENTS = {
    "simple":    SimpleAgent,
    "greedy":    GreedySwitchAgent,
    "stalker":   StalkerAgent,
    "chameleon": ChameleonAgent,
}
opponent_agent = _OPPONENTS[OPPONENT](1, map_layout=map_layout)
opponent_name  = opponent_agent.__class__.__name__
map_name = MAP_NAME
om_id = f"{MAP_NAME.lower()}_{OPPONENT}"
classic_id = om_id

# Environment + args (architecture params must match training so state_dicts load)
env = SimpleForagingEnv(max_steps=100, map_layout=map_layout)
obs_sample = env.reset()[0]
H, W, F_dim = obs_sample.shape
NUM_ACTIONS = 4

args = OMGArgs(
    device=device,
    folder_id=FOLDER_ID,
    batch_size=128,
    qnet_hidden=256,
    cnn_hidden=64,
    max_steps=50,
    tau_start=1.5,
    tau_end=0.1,
    tau_decay_steps=600_000,
    state_shape=obs_sample.shape,
    H=H, W=W,
    action_dim=NUM_ACTIONS,
    d_model=128,
    nhead=4,
    num_encoder_layers=2,
    dim_feedforward=256,
    dropout=0.05,
)

# Build agents
inference_model = t.SpatialOpponentModel(args=args).to(args.device)
op_model = OpponentModel(model=inference_model, args=args)
q_agent = QLearningAgent(env=env, opponent_model=op_model, args=args)
q_agent_classic = QLearningAgentClassic(env=env, args=args)


# ---------------------------------------------------------------- checkpoint loading
def _resolve_ckpt(folder, prefix):
    # Return the highest-epoch *_ep*.pth for a prefix, or None.
    files = glob.glob(f"./models/{folder}/{prefix}_ep*.pth")
    if not files:
        return None

    def _epoch_num(f):
        try:
            return int(f.split("_ep")[-1].split(".")[0])
        except ValueError:
            return -1

    return max(files, key=_epoch_num)


def _load(prefix, net, optional=False):
    path = None
    if EPOCH is not None:
        cand = f"./models/{FOLDER_ID}/{prefix}_ep{EPOCH}.pth"
        if os.path.exists(cand):
            path = cand
    if path is None:
        path = _resolve_ckpt(FOLDER_ID, prefix)
    if path is None:
        path = f"./models/{FOLDER_ID}/{prefix}.pth"  # legacy unsuffixed name
    if not os.path.exists(path):
        msg = f"Checkpoint not found for '{prefix}' (looked in ./models/{FOLDER_ID}/)."
        if optional:
            print(f"[skip] {msg}")
            return
        raise FileNotFoundError(msg)
    net.load_state_dict(torch.load(path, map_location=args.device))
    print(f"Loaded {prefix}: {path}")


_load("om_qnet", q_agent.q)
_load("om_inference", q_agent.model.inference_model)
_load("classic_qnet", q_agent_classic.q)

q_agent.q.eval()
q_agent.model.inference_model.eval()
q_agent_classic.q.eval()
print(f"\nReady: {MAP_NAME} vs {opponent_name} | device={device}")
```

```python
# Functions to run / evaluate episodes with the agents.
# These mirror QLearningAgent.run_test_episode but add OM-heatmap rendering,
# KL/spatial error and Q-value sensitivity metrics.


def run_om_episode(env, q_agent, opponent_agent, render=False, max_steps=None):
    q_agent.model.inference_model.eval()
    max_steps = max_steps or q_agent.args.max_steps
    device = q_agent.args.device

    obs = env.reset()
    opponent_agent.reset()
    q_agent.tracker.reset(use_map_prior=q_agent.args.belief_map_prior)
    q_agent.tracker.update(obs[0])

    done = False
    ep_ret = 0.0
    opp_ret = 0.0
    ep_entropy = 0.0
    ep_kl_errors = []
    ep_spatial_errors = []
    ep_opp_rewards = []
    ep_sensitivity_ratios = []
    ep_pred_vs_true = []
    ep_true_vs_uniform = []

    # Transformer history (cached state features + motion encoding via prev_obs)
    history_len = q_agent.args.max_history_length
    rolling_feats = torch.zeros((1, history_len, q_agent.args.d_model), device=device)
    rolling_mask = torch.zeros((1, history_len), dtype=torch.bool, device=device)
    current_seq_len = 0
    prev_state_tensor = torch.zeros((1, *obs[0].shape), device=device)

    step = 0
    for step in range(max_steps):
        history = {
            "state_features": rolling_feats,
            "mask": rolling_mask,
            "prev_obs": prev_state_tensor,
        }
        s_aug = q_agent.tracker.augment(obs[0])

        a, g_map, step_entropy = q_agent.select_action(
            obs[0], s_aug, history, eval=True
        )
        a_opponent, _, opp_heatmap = opponent_agent.select_action(obs[1], eval=True)
        actions = {0: a, 1: a_opponent}

        # Q-value sensitivity to the inferred subgoal (uses augmented state)
        sens = g_map_sensitivity_full(q_agent, obs[0], s_aug, history, opp_heatmap)
        pvt = sens["sensitivity_pred_vs_true"]
        tvu = sens["sensitivity_true_vs_uniform"]
        ratio = pvt / tvu if tvu > 0.05 else None
        ep_sensitivity_ratios.append(ratio)
        ep_pred_vs_true.append(pvt)
        ep_true_vs_uniform.append(tvu)

        if render:
            print(f"Step {step} | Return {ep_ret:.2f}")
            q_agent.heatmap_subgoal(g_map, save=False)
            time.sleep(0.2)
            env.render_from_obs(obs[0])
            print(
                f"Sensitivity (Pred vs Uniform): {sens['sensitivity_pred_vs_uniform']:.4f} | "
                f"(Pred vs True): {pvt:.4f} | (True vs Uniform): {tvu:.4f}"
            )

        if g_map.dim() == 2:
            g_map = g_map.unsqueeze(0)  # (1, H, W)
        opp_heatmap_t = torch.from_numpy(opp_heatmap).unsqueeze(0).to(device)

        kl_error = q_agent.model.heatmap_kl_divergence(g_map, opp_heatmap_t)
        spatial_error = q_agent.model.expected_spatial_error(g_map, opp_heatmap_t)
        ep_kl_errors.append(kl_error)
        ep_spatial_errors.append(spatial_error)

        next_obs, reward, done, info = env.step(actions)
        q_agent.tracker.update(next_obs[0])

        # Cache the state feature for the next history window
        state_tensor = torch.from_numpy(obs[0]).float().unsqueeze(0).to(device)
        with torch.no_grad():
            new_feat = q_agent.model.inference_model.get_features(
                state_tensor, prev_state_tensor
            )

        rolling_feats = torch.roll(rolling_feats, shifts=-1, dims=1)
        rolling_mask = torch.roll(rolling_mask, shifts=-1, dims=1)
        rolling_feats[:, -1, :] = new_feat
        if current_seq_len < history_len:
            current_seq_len += 1
        rolling_mask[:, -current_seq_len:] = True
        prev_state_tensor = state_tensor

        ep_ret += reward[0]
        opp_ret += reward[1]
        ep_opp_rewards.append(reward[1])
        obs = next_obs
        ep_entropy += step_entropy
        if done:
            break

    # Post-episode trimming: ignore steps after the opponent already failed
    kd_errors, spatial_errors = [], []
    if opp_ret == 0 and ep_ret > 0:
        last_valid_step = len(ep_opp_rewards)
        for t in reversed(range(len(ep_opp_rewards))):
            if ep_opp_rewards[t] > 0:
                last_valid_step = t + 1
                break
        if last_valid_step > 0:
            kd_errors.extend(ep_kl_errors[:last_valid_step])
            spatial_errors.extend(ep_spatial_errors[:last_valid_step])
    else:
        kd_errors.extend(ep_kl_errors)
        spatial_errors.extend(ep_spatial_errors)

    return {
        "return": ep_ret,
        "steps": step + 1,
        "opp_return": opp_ret,
        "avg_entropy": ep_entropy / (step + 1),
        "avg_kl_error": np.mean(kd_errors) if kd_errors else None,
        "avg_spatial_error": np.mean(spatial_errors) if spatial_errors else None,
        "ratio_sensitivity_pred_true_to_true_uniform": ep_sensitivity_ratios,
        "pred_vs_true_sensitivity": ep_pred_vs_true,
        "true_vs_uniform_sensitivity": ep_true_vs_uniform,
    }


def run_classic_episode(env, q_agent_classic, opponent_agent, render=False, max_steps=None):
    q_agent_classic.q.eval()
    max_steps = max_steps or q_agent_classic.args.max_steps

    obs = env.reset()
    opponent_agent.reset()
    q_agent_classic.tracker.reset(use_map_prior=q_agent_classic.args.belief_map_prior)
    q_agent_classic.tracker.update(obs[0])

    done = False
    ep_ret = 0.0
    ep_entropy = 0.0
    step = 0
    for step in range(max_steps):
        s_aug = q_agent_classic.tracker.augment(obs[0])
        a, step_entropy = q_agent_classic.select_action(s_aug, eval=True)
        a_opponent, _, _ = opponent_agent.select_action(obs[1], eval=True)
        actions = {0: a, 1: a_opponent}

        if render:
            clear_output(wait=True)
            print(f"Step {step} | Return {ep_ret:.2f}")
            q_agent_classic.heatmap_q_values(save=False)
            time.sleep(0.2)
            env.render_from_obs(obs[0])

        next_obs, reward, done, info = env.step(actions)
        q_agent_classic.tracker.update(next_obs[0])
        ep_ret += reward[0]
        obs = next_obs
        ep_entropy += step_entropy
        if done:
            break
    return {
        "return": ep_ret,
        "steps": step + 1,
        "avg_entropy": ep_entropy / (step + 1),
    }


@torch.no_grad()
def g_map_sensitivity_full(q_agent, obs, s_aug, history, true_heatmap):
    # How much does Q care about the inferred subgoal heatmap?
    # Compares Q(s, g_pred) against Q(s, g_true) and Q(s, g_uniform).
    # Uses the belief-augmented state for the Q-network input.
    device = q_agent.args.device
    x = torch.from_numpy(obs).float().unsqueeze(0).to(device)        # raw obs -> OM
    x_aug = torch.from_numpy(s_aug).float().unsqueeze(0).to(device)  # augmented -> Q

    g_logits = q_agent.model(x, history)
    g_pred = F.softmax(g_logits.view(1, -1), dim=-1).view_as(g_logits)

    g_true = torch.from_numpy(true_heatmap).float().unsqueeze(0).to(device)
    g_true = g_true / g_true.sum().clamp(min=1e-8)

    # Uniform baseline over non-wall cells (channel 0 == empty)
    valid_mask = torch.from_numpy(obs[:, :, 0]).float().unsqueeze(0).to(device)
    g_uniform = valid_mask / valid_mask.sum().clamp(min=1e-8)

    q_pred = q_agent.q(x_aug, g_pred)
    q_true = q_agent.q(x_aug, g_true)
    q_uniform = q_agent.q(x_aug, g_uniform)

    return {
        "sensitivity_pred_vs_uniform": (q_pred - q_uniform).abs().mean().item(),
        "sensitivity_pred_vs_true": (q_pred - q_true).abs().mean().item(),
        "sensitivity_true_vs_uniform": (q_true - q_uniform).abs().mean().item(),
    }


@torch.no_grad()
def _final_memory(model, x, history):
    # Replicates SpatialOpponentModel.forward but returns the memory token (B, d_model).
    model.eval()
    B, H, W, F_dim = x.shape
    hist_mask = history["mask"]
    hist_feat = history["state_features"]
    x_prev = history.get("prev_obs", torch.zeros_like(x))

    x_feat = model.get_features(x, x_prev).unsqueeze(1)
    seq_feats = torch.cat([x_feat, hist_feat], dim=1)

    x_mask = torch.ones((B, 1), dtype=torch.bool, device=x.device)
    full_mask = torch.cat([x_mask, hist_mask], dim=1)

    seq_feats = seq_feats * np.sqrt(model.args.d_model)
    seq_feats = model.pos_encoder(seq_feats)
    memory = model.transformer(seq_feats, src_key_padding_mask=~full_mask)
    return memory[:, 0, :]


@torch.no_grad()
def visualize_final_memory_3d(model, x, history, target_map):
    model.eval()
    B, H, W, F_dim = x.shape
    final_memory = _final_memory(model, x, history)
    final_memory_np = final_memory.cpu().numpy()

    true_goals_flat = [np.argmax(b.cpu().numpy()) for b in target_map]
    valid_indices, readable_goals = [], []
    for i, g in enumerate(true_goals_flat):
        if target_map[i].cpu().numpy().max() > 0:
            r, c = g // W, g % W
            if x[i, r, c, 1] == 1.0:
                readable_goals.append(f"Food at ({r}, {c})")
                valid_indices.append(i)

    if not valid_indices:
        print("No valid food targets found in this batch to visualize.")
        return

    final_memory_filtered = final_memory_np[valid_indices]
    tsne = TSNE(n_components=3, perplexity=30, random_state=42)
    memory_3d = tsne.fit_transform(final_memory_filtered)

    df = pd.DataFrame({
        "t-SNE 1": memory_3d[:, 0],
        "t-SNE 2": memory_3d[:, 1],
        "t-SNE 3": memory_3d[:, 2],
        "Target Subgoal": readable_goals,
    })
    fig = px.scatter_3d(
        df, x="t-SNE 1", y="t-SNE 2", z="t-SNE 3",
        color="Target Subgoal",
        title="3D t-SNE of Transformer Belief State",
        opacity=0.8,
    )
    fig.update_traces(marker=dict(size=5))
    fig.show()


@torch.no_grad()
def visualize_final_memory_2d(model, x, history, target_map, filename="model_beliefs.png"):
    model.eval()
    B, H, W, F_dim = x.shape
    final_memory = _final_memory(model, x, history)
    final_memory_np = final_memory.cpu().numpy()

    true_goals_flat = [np.argmax(b.cpu().numpy()) for b in target_map]
    valid_indices, readable_goals = [], []
    for i, g in enumerate(true_goals_flat):
        if target_map[i].cpu().numpy().max() > 0:
            r, c = g // W, g % W
            if x[i, r, c, 1] == 1.0:
                readable_goals.append(f"Food at ({r}, {c})")
                valid_indices.append(i)

    if not valid_indices:
        print("No valid food targets found in this batch to visualize.")
        return

    final_memory_filtered = final_memory_np[valid_indices]
    tsne = TSNE(n_components=2, perplexity=30, random_state=42)
    memory_2d = tsne.fit_transform(final_memory_filtered)

    plt.figure(figsize=(10, 7))
    unique_labels = sorted(set(readable_goals))
    cmap = plt.get_cmap("tab10")
    shapes = ["o", "s", "D", "^", "v", "<", ">", "P", "X", "*"]

    for i, label in enumerate(unique_labels):
        idx = [j for j, l in enumerate(readable_goals) if l == label]
        color = cmap(i % 10)
        marker = shapes[(i // 10) % len(shapes)]
        plt.scatter(
            memory_2d[idx, 0], memory_2d[idx, 1],
            c=[color], marker=marker, label=label,
            alpha=0.8, s=60, edgecolors="w", linewidths=0.5,
        )

    plt.title("2D t-SNE of Transformer Belief State", fontsize=16, pad=15)
    plt.xlabel("t-SNE Dimension 1", fontsize=12)
    plt.ylabel("t-SNE Dimension 2", fontsize=12)
    plt.legend(title="True Subgoal Location", title_fontsize=13, fontsize=11,
               bbox_to_anchor=(1.05, 1), loc="upper left", frameon=True)
    plt.grid(True, linestyle="--", alpha=0.5)
    plt.tight_layout()
    plt.savefig(filename, dpi=300, bbox_inches="tight")
    plt.show()
    plt.close()


def collect_data(env, agent0, agent1, num_episodes=500):
    # Collects transitions (with hindsight true_goal_map) using the OM agent.
    # Each transition stores cached state features + prev_obs so the batch can be
    # fed to the OM with cached_features=True.
    from collect_data import _apply_hindsight_relabeling

    master_dataset = []
    H, W, _ = env.reset()[0].shape
    for _ in range(num_episodes):
        obs = env.reset()
        agent0.reset()
        agent1.reset()
        agent0.tracker.reset(use_map_prior=agent0.args.belief_map_prior)
        agent0.tracker.update(obs[0])

        episode_transitions = []
        history_len = agent0.args.max_history_length
        rolling_feats = torch.zeros((1, history_len, agent0.args.d_model), device=agent0.args.device)
        rolling_mask = torch.zeros((1, history_len), dtype=torch.bool, device=agent0.args.device)
        current_seq_len = 0
        prev_state_tensor = torch.zeros((1, *obs[0].shape), device=agent0.args.device)

        for step in range(agent0.args.max_steps):
            history = {
                "state_features": rolling_feats,
                "mask": rolling_mask,
                "prev_obs": prev_state_tensor,
            }
            s_aug = agent0.tracker.augment(obs[0])
            a_0, _, _ = agent0.select_action(obs[0], s_aug, history, eval=True)
            a_1, _, true_opp_heatmap = agent1.select_action(obs[1])
            actions = {0: a_0, 1: a_1}

            global_state = env.get_global_state()
            next_obs, reward, done, info = env.step(actions)
            next_global_state = env.get_global_state()
            agent0.tracker.update(next_obs[0])

            transition = {
                "state": obs[0].copy(),
                "global_state": global_state.copy(),
                "reward": float(reward[0]),
                "opp_reward": float(reward[1]),
                "next_state": next_obs[0].copy(),
                "next_global_state": next_global_state.copy(),
                "done": bool(done),
                "true_opp_heatmap": true_opp_heatmap.copy(),
                "hist_len": len(episode_transitions),
                "history": {
                    "state_features": rolling_feats.squeeze(0).cpu().numpy().copy(),
                    "mask": rolling_mask.squeeze(0).cpu().numpy().copy(),
                    "prev_obs": prev_state_tensor.squeeze(0).cpu().numpy().copy(),
                },
            }
            episode_transitions.append(transition)

            state_tensor = torch.from_numpy(obs[0]).float().unsqueeze(0).to(agent0.args.device)
            with torch.no_grad():
                new_feat = agent0.model.inference_model.get_features(state_tensor, prev_state_tensor)
            rolling_feats = torch.roll(rolling_feats, shifts=-1, dims=1)
            rolling_mask = torch.roll(rolling_mask, shifts=-1, dims=1)
            rolling_feats[:, -1, :] = new_feat
            if current_seq_len < history_len:
                current_seq_len += 1
            rolling_mask[:, -current_seq_len:] = True
            prev_state_tensor = state_tensor

            obs = next_obs
            if done:
                break

        _apply_hindsight_relabeling(episode_transitions, H, W)
        master_dataset.extend(episode_transitions)
    return master_dataset
```

```python
# NOTE: 10_000 episodes x 3 runs is slow on CPU -- lower total_episodes/num_runs for a quick check.
num_runs = 3
total_episodes = 10_000
avg_per = 1000
num_bins = total_episodes // avg_per

# Matrices to hold the binned averages for all runs
# Shape will be (3 runs, 10 bins)
all_runs_avg_returns = np.zeros((num_runs, num_bins))
all_runs_avg_spatial_errors = np.zeros((num_runs, num_bins))

for run in range(num_runs):
    print(f"--- Starting Evaluation Run {run + 1}/{num_runs} ---")

    episode_returns = []
    spatial_errors = []

    run_avg_returns = []
    run_avg_spatial_errors = []

    for episode in range(total_episodes):
        results = run_om_episode(env, q_agent, opponent_agent, render=False)
        episode_returns.append(results["return"])

        # Handle potential None values safely
        if results["avg_spatial_error"] is not None:
            spatial_errors.append(results["avg_spatial_error"])

        # Calculate bins every `avg_per` episodes
        if (episode + 1) % avg_per == 0:
            avg_return = np.mean(episode_returns[-avg_per:])

            # Safe mean calculation for spatial errors
            if len(spatial_errors) > 0:
                avg_spatial_error = np.mean(
                    spatial_errors[-min(len(spatial_errors), avg_per):])
            else:
                avg_spatial_error = 0.0

            run_avg_returns.append(avg_return)
            run_avg_spatial_errors.append(avg_spatial_error)

    # Store this run's data into the master matrices
    all_runs_avg_returns[run] = run_avg_returns
    all_runs_avg_spatial_errors[run] = run_avg_spatial_errors


x_axis = np.arange(1, num_bins + 1)

mean_returns = np.mean(all_runs_avg_returns, axis=0)
std_returns = np.std(all_runs_avg_returns, axis=0)

mean_spatial = np.mean(all_runs_avg_spatial_errors, axis=0)
std_spatial = np.std(all_runs_avg_spatial_errors, axis=0)

# --- 1. Plot Returns ---
plt.figure(figsize=(10, 5))
# Plot the solid mean line
plt.plot(x_axis, mean_returns,
         label=f'Mean Return', color='blue', linewidth=2)
# Shade the variance band
plt.fill_between(x_axis,
                 mean_returns - std_returns,
                 mean_returns + std_returns,
                 color='blue', alpha=0.2, label='Std Dev')

plt.xlabel('1e3 Episodes')
plt.ylabel('Return')
plt.title(f'OM-DDQN Agent vs {opponent_name} {map_name}')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.savefig(f'{om_id}_{opponent_name}_OM.png',
            dpi=300, bbox_inches='tight')
plt.show()

# --- 2. Plot Spatial Errors ---
plt.figure(figsize=(10, 5))
# Plot the solid mean line
plt.plot(x_axis, mean_spatial,
         label=f'Mean Spatial Error', color='red', linewidth=2)
# Shade the variance band
plt.fill_between(x_axis,
                 mean_spatial - std_spatial,
                 mean_spatial + std_spatial,
                 color='red', alpha=0.2, label='Std Dev')

plt.xlabel('1e3 Episodes')
plt.ylabel('Average Spatial Error (tiles)')
plt.title(f'Agent with OM vs {opponent_name} {map_name} - Spatial Error')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.savefig(f'{om_id}_{opponent_name}_spatial_error.png',
            dpi=300, bbox_inches='tight')
plt.show()
```

```python
# NOTE: 10_000 episodes is slow on CPU -- lower total_episodes for a quick check.
total_episodes = 1_000
avg_per = 100
max_steps = q_agent.args.max_steps  # 50

avg_returns = []
episode_returns = []
spatial_errors = []
avg_spatial_errors = []

# --- NEW: Buckets for step-by-step metrics ---
# Create max_steps empty lists to hold the values for each specific timestep
step_ratios_data = [[] for _ in range(max_steps)]
step_pvt_data = [[] for _ in range(max_steps)]
step_tvu_data = [[] for _ in range(max_steps)]

for episode in range(total_episodes):
    results = run_om_episode(env, q_agent, opponent_agent, render=False)

    # Standard episode-level tracking
    episode_returns.append(results["return"])
    if results["avg_spatial_error"] is not None:
        spatial_errors.append(results["avg_spatial_error"])

    if (episode + 1) % avg_per == 0:
        avg_returns.append(np.mean(episode_returns[-avg_per:]))
        avg_spatial_errors.append(np.mean(spatial_errors[-min(len(spatial_errors), avg_per):]))

    # --- NEW: Distribute step-level metrics into their buckets ---
    ratios = results["ratio_sensitivity_pred_true_to_true_uniform"]
    pvt = results["pred_vs_true_sensitivity"]
    tvu = results["true_vs_uniform_sensitivity"]

    # enumerate gives us the step index (0, 1, 2...) safely up to the episode's actual length
    for step_idx, ratio_val in enumerate(ratios):
        if ratio_val is not None:
            step_ratios_data[step_idx].append(ratio_val)

    for step_idx, pvt_val in enumerate(pvt):
        if pvt_val is not None:
            step_pvt_data[step_idx].append(pvt_val)

    for step_idx, tvu_val in enumerate(tvu):
        if tvu_val is not None:
            step_tvu_data[step_idx].append(tvu_val)

print("Evaluation Complete. Processing step-by-step averages...")

# --- Process the Buckets ---
valid_steps = []
avg_ratios = []
avg_pvt = []
avg_tvu = []
surviving_episodes = [] # Good to know how many episodes made it to this step

# We require at least 50 episodes to reach a step to plot it, to avoid massive variance noise
MIN_SAMPLES = 50

for i in range(max_steps):
    if len(step_ratios_data[i]) >= MIN_SAMPLES:
        valid_steps.append(i)
        avg_ratios.append(np.mean(step_ratios_data[i]))
        avg_pvt.append(np.mean(step_pvt_data[i]))
        avg_tvu.append(np.mean(step_tvu_data[i]))
        surviving_episodes.append(len(step_ratios_data[i]))

# --- Plotting ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: The Sensitivities (Pred vs True AND True vs Uniform)
ax1.plot(valid_steps, avg_pvt, label="Prediction Error Cost (Pred vs True)", color="red", linewidth=2)
ax1.plot(valid_steps, avg_tvu, label="Max Utility (True vs Uniform)", color="blue", linewidth=2)
ax1.set_title("Absolute Q-Value Sensitivity over Time")
ax1.set_xlabel("Episode Step")
ax1.set_ylabel("Absolute Q-Value Difference")
ax1.grid(True, alpha=0.3)
ax1.legend()

# Plot 2: The Ratio
# Create a twin axis so we can overlay the survival count (how many games lasted this long)
ax2_twin = ax2.twinx()
ax2_twin.fill_between(valid_steps, 0, surviving_episodes, color="gray", alpha=0.15, label="Surviving Episodes")
ax2_twin.set_ylabel("Number of Surviving Episodes", color="gray")
ax2_twin.tick_params(axis='y', labelcolor="gray")

ax2.plot(valid_steps, avg_ratios, label="Sensitivity Ratio", color="purple", linewidth=2.5)
ax2.set_title("Sensitivity Ratio (Pred/True Error vs Max Utility)")
ax2.set_xlabel("Episode Step")
ax2.set_ylabel("Ratio")
ax2.grid(True, alpha=0.3)

# Combine legends for Plot 2
lines, labels = ax2.get_legend_handles_labels()
lines2, labels2 = ax2_twin.get_legend_handles_labels()
ax2.legend(lines + lines2, labels + labels2, loc="upper right")

plt.tight_layout()
plt.savefig(f"{om_id}_sensitivity.png", dpi=300)
plt.show()
```

```python
from maps import MAP_1, MAP_2, MAP_3, MAP_4, MAP_5

maps_dict = {"MAP_1": MAP_1,"MAP_2": MAP_2,"MAP_3": MAP_3, "MAP_4": MAP_4, "MAP_5": MAP_5}

# --- 2. Visualization Function ---
def generate_map(map_data, title, filename):
  # Mapping characters to integers for the colormap
  # 0: Empty, 1: Wall, 2: Food, 3: Agent A, 4: Agent B
  char_to_num = {'.': 0, '#': 1, 'o': 2, 'A': 3, 'B': 4}

  height = len(map_data)
  width = len(map_data[0])
  grid = np.zeros((height, width), dtype=int)

  for y, row in enumerate(map_data):
      for x, char in enumerate(row):
          grid[y, x] = char_to_num.get(char, 0)


  colors = ['#FFFFFF', '#555555', '#2CA02C', '#1F77B4', '#D62728']
  cmap = mcolors.ListedColormap(colors)
  bounds = [-0.5, 0.5, 1.5, 2.5, 3.5, 4.5]
  norm = mcolors.BoundaryNorm(bounds, cmap.N)

  fig, ax = plt.subplots(figsize=(width * 0.4, height * 0.4 + 1))

  ax.imshow(grid, cmap=cmap, norm=norm)

  for y, row in enumerate(map_data):
      for x, char in enumerate(row):
          if char in ['A', 'B']:
              ax.text(x, y, char, ha='center', va='center',
                      color='white', fontweight='bold', fontsize=12)

  ax.set_xticks(np.arange(-.5, width, 1), minor=True)
  ax.set_yticks(np.arange(-.5, height, 1), minor=True)
  ax.grid(which='minor', color='black', linestyle='-', linewidth=1.5)


  ax.tick_params(which='both', bottom=False, left=False,
                  labelbottom=False, labelleft=False)


  legend_elements = [
      mpatches.Patch(facecolor=colors[0], edgecolor='black', label='Empty'),
      mpatches.Patch(facecolor=colors[1], edgecolor='black', label='Wall'),
      mpatches.Patch(facecolor=colors[2], edgecolor='black', label='Food'),
      mpatches.Patch(facecolor=colors[3], edgecolor='black', label='Agent A'),
      mpatches.Patch(facecolor=colors[4], edgecolor='black', label='Agent B')
  ]


  ax.legend(handles=legend_elements, loc='upper center',
            bbox_to_anchor=(0.5, -0.05), ncol=5, frameon=False, fontsize=10)

  plt.title(title, fontsize=16, pad=15)

  plt.savefig(filename, format='pdf', dpi=300, bbox_inches='tight')
  plt.close()
  print(f"Saved: {filename}")

for map_name, map_layout in maps_dict.items():
    clean_title = map_name.replace("_", " ").title()
    generate_map(map_layout, title=clean_title, filename=f"{map_name}.pdf")
```
